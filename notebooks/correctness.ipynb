# Correctness on xCodeEval

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')

df = pd.read_json('../results/trials/results_codescope_test_generation_java_with_gemini-2.0-flash_seed42_20250901_191356.json')

plt.figure(figsize=(10, 6))
sns.lineplot(data=df, x=df.index, y='branch_cov_orig', label='Originals', color='#2a9d8f')
sns.lineplot(data=df, x=df.index, y='branch_cov_span', label='Variants', color='#e76f51')
plt.axhline(y=df['branch_cov_span_avg'].mean(), color='#e9c46a', linestyle='--', label='Avg. of Variants')

plt.title('Test Generation on CodeScope', fontsize=16)
plt.xlabel('Sequence', fontsize=12)
plt.ylabel('Branch Coverage', fontsize=12)
plt.ylim(.8, 1.0)
plt.legend()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

df_code = pd.read_csv('../results/correctness_xCodeEval_code2tag_java_with_gemini-2.0-flash_20250726_115827.csv')
df_des_code = pd.read_csv('../results/correctness_xCodeEval_des_code2tag_java_with_gemini-2.0-flash_20250726_114229.csv')

# for code2tag
plt.figure(figsize=(10, 6))
sns.lineplot(data=df_code, x=df_code.index, y=df_code.columns[0], label='Originals', color='#2a9d8f')
sns.lineplot(data=df_code, x=df_code.index, y=df_code.columns[1], label='Variants', color='#e76f51')
avg_value = df_code[df_code.columns[1]].mean()
plt.axhline(y=avg_value, color='#e9c46a', linestyle='--', label='Avg. of Variants')

plt.title('Code2Tag on xCodeEval', fontsize=16)
plt.xlabel('Sequence', fontsize=12)
plt.ylabel('Macro F1 Score', fontsize=12)
plt.ylim(.35, .8)
plt.legend()

# for des_code2tag
plt.figure(figsize=(10, 6))
sns.lineplot(data=df_des_code, x=df_des_code.index, y=df_des_code.columns[0], label='Originals', color='#2a9d8f')
sns.lineplot(data=df_des_code, x=df_des_code.index, y=df_des_code.columns[1], label='Variants', color='#e76f51')
avg_value = df_des_code[df_des_code.columns[1]].mean()
plt.axhline(y=avg_value, color='#e9c46a', linestyle='--', label='Avg. of Variants')

plt.title('DesCode2Tag on xCodeEval', fontsize=16)
plt.xlabel('Sequence', fontsize=12)
plt.ylabel('Macro F1 Score', fontsize=12)
plt.ylim(.35, .8)
plt.legend()

plt.show()
```
